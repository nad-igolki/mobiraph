```python
import os

def get_insect_names(folder_path):
    names = set()

    for filename in os.listdir(folder_path):
        parts = filename.split('_')

        if parts[0] in {"anopheles", "insect"}:
            name = "_".join(parts[:3]) if parts[0] == "insect" else "_".join(parts[:2])
            name = name.removeprefix("insect_")
            names.add(name)

    return sorted(names)


folder = "/Users/nad/Downloads/svs_insect_nadya_2026_01_04/"
insects = set(get_insect_names(folder))
print(get_insect_names(folder))
```

```python
import json

def get_animals_from_json(path):
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    return sorted({
        v["organism"].lower().replace(" ", "_")
        for v in data.values()
        if "organism" in v
    })


animals = set(get_animals_from_json("/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/metadata_02.json"))
print(animals)
```

```python
with open("/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/animals.txt", "w", encoding="utf-8") as f:
    for a in sorted(animals):
        f.write(a + "\n")
```

```python
import os

def get_plant_organisms(directory):
    organisms = []

    for filename in os.listdir(directory):
        if filename.startswith("plant_"):
            name_part = filename[len("plant_"):]

            organism = os.path.splitext(name_part)[0]
            organisms.append(organism)

    return organisms


directory_path = "/Users/nad/hse/semester08/mobiraph/data/n18_other_organisms"
result = get_plant_organisms(directory_path)

print(result)
```

```python
import json

def get_ids_from_json(path):
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    ids = []

    for id, value in data.items():
        if value["organism"].lower().replace(" ", "_") in intersection:
            ids.append(id)

    return ids


intersection_ids = set(get_ids_from_json("/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/metadata_02.json"))
print(intersection_ids)
```

```python
with open("/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/intersection_plants_ids.txt", "w", encoding="utf-8") as f:
    for a in sorted(intersection_ids):
        f.write(a + "\n")
```

```python
ids_file = "/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/intersection_plants_ids.txt"
input_fasta = "/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/all_sequences_filtered_02.fasta"
output_fasta = "/Users/nad/hse/semester08/mobiraph/data/n13_repbase_processed/intersection_plants.fasta"

with open(ids_file) as f:
    wanted_ids = set(line.strip() for line in f)

with open(input_fasta) as fin, open(output_fasta, "w") as fout:
    write = False

    for line in fin:
        if line.startswith(">"):
            header = line[1:].strip()
            seq_id = header.split()[0]

            write = seq_id in wanted_ids

        if write:
            fout.write(line)
```
