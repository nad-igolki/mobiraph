```python
mapping = {
    'Bel-Pao': 'BEL',
    'CACTA': 'EnSpm/CACTA',
    'Copia': 'Copia',
    'Crypton': 'DNA transposon_other',
    'DIRS': 'DIRS',
    'Gypsy': 'Gypsy',
    'Helitron': 'Helitron',
    'I': 'Non-LTR Retrotransposon_other',
    'Jockey': 'Non-LTR Retrotransposon_other',
    'L1': 'L1',
    'Merlin': 'DNA transposon_other',
    'Mutator': 'MuDR',
    'P': 'DNA transposon_other',
    'PIF-Harbinger': 'Harbinger',
    'Penelope': 'Non-LTR Retrotransposon_other',
    'R2': 'Non-LTR Retrotransposon_other',
    'RTE': 'RTE',
    'Retrovirus': 'Gypsy',
    'Tc1-Mariner': 'Mariner/Tc1',
    'Transib': 'DNA transposon_other',
    'Unknown': 'Unknown',
    'hAT': 'hAT',
    'tRNA': 'Unknown'
}
```

```python
name_mapping = {}
with open("/Users/nad/mobiraph/data/n13_repbase_processed/all_sequences_ltr_correction_test_mapping.csv") as f:
    next(f)
    for line in f:
        new, old = line.strip().split("\t")
        name_mapping[new] = old
```

```python
import csv

neuralte_classification_result = {}

with open('/Users/nad/mobiraph/neuralte_test_result_unified/classified.info', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        key = row['#Seq_Name']
        value = row['Predict_Label']
        neuralte_classification_result[name_mapping[key]] = mapping[value]
```

```python
filtered_metadata = {}
for key, elem_dict in metadata.items():
    if 'superfamily' in elem_dict:
        filtered_metadata[key] = elem_dict['superfamily']
```

```python
import csv

mobiraph_result = {}

with open('/Users/nad/mobiraph/mobiraph_prediction2.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        key = row['name']
        value = row['prediction']
        mobiraph_result[key] = value
```

```python
common = set(filtered_metadata.keys()) & set(neuralte_classification_result.keys()) & set(mobiraph_result.keys())
```

```python
len(common)
```

```python
true = []
pred = []
for seq_id in common:
    true.append(filtered_metadata[seq_id])
    pred.append(neuralte_classification_result[seq_id])
```

```python
from sklearn.metrics import classification_report

report = classification_report(true, pred)
print(report)
```

```python
mtrue = []
mpred = []
for seq_id in common:
    mtrue.append(filtered_metadata[seq_id])
    mpred.append(mobiraph_result[seq_id])
```

```python
from sklearn.metrics import classification_report

report = classification_report(mtrue, mpred)
print(report)
```
